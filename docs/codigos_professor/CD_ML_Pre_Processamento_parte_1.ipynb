#3.Imputação de Valores

##3.1 Preenchimento de valores ausentes com a **média**

````python
# Criar um DataFrame de exemplo com valores ausentes (NaN)
import pandas as pd
import numpy as np
data = {'janeiro': [1, 2, np.nan, 4, 5],
        'fevereiro': [np.nan, 6, 7, 8, np.nan],
        'março': [10, 11, 12, np.nan, 14]}
df = pd.DataFrame(data)

print("\nNúmero de Veiculos Vendidos")
print(df.isnull())

print("\nDataFrame original:")
print("Número de Veiculos Vendidos")
print(df)
print("\n")

# Calcular a média de cada coluna
media_coluna_jan = df['janeiro'].mean()
media_coluna_fev = df['fevereiro'].mean()
media_coluna_mar = df['março'].mean()

print(f"Média da coluna A: {media_coluna_jan:.2f}")
print(f"Média da coluna B: {media_coluna_fev:.2f}")
print(f"Média da coluna C: {media_coluna_mar:.2f}")
print("\n")

# Preencher os valores ausentes na coluna 'A' com sua média
# df['janeiro'].fillna(media_coluna_jan, inplace=True)
df.fillna({'janeiro':media_coluna_jan}, inplace=True)

# Preencher os valores ausentes na coluna 'B' com sua média
# df['fevereiro'].fillna(media_coluna_fev, inplace=True)
df['fevereiro'] = df['fevereiro'].fillna(media_coluna_fev)

# Preencher os valores ausentes na coluna 'C' com sua média
# df['março'].fillna(media_coluna_mar, inplace=True)
df.fillna({'março':media_coluna_mar}, inplace=True)

print("DataFrame após preencher os valores ausentes com a média de cada coluna:")
print("\nNúmero de Veiculos Vendidos")
print(df)
print()
````

##2.4 Interpolação (séries temporais): preenchimento com o **anterior/posterior**


````python
import pandas as pd
import numpy as np

# Exemplo de dados temporais com valores ausentes
data = {
    'data': pd.date_range('2024-01-01', periods=7),
    'temperatura': [22.1, np.nan, np.nan, 23.0, np.nan, 24.5, np.nan]
}
df = pd.DataFrame(data)
print()
print(df.info())
print()

print("Original:")
print(df)

# Preenchimento com valor anterior (forward fill)
print("\nForward Fill (valor anterior):")
df_ffill = df.copy()
# df_ffill['temperatura'] = df_ffill['temperatura'].fillna(method='ffill')
df_ffill['temperatura'] = df_ffill['temperatura'].ffill()
print(df_ffill)

# Preenchimento com valor posterior (backward fill)
print("\nBackward Fill (valor posterior):")
df_bfill = df.copy()
# df_bfill['temperatura'] = df_bfill['temperatura'].fillna(method='bfill')
df_bfill['temperatura'] = df_bfill['temperatura'].bfill()
print(df_bfill)



````

# 5.Transformação de Dados (Parte 1)

A **transformação de dados** é uma etapa do pré-processamento em que as **variáveis originais** são **modificadas**, **reescalonadas** ou **convertidas** para uma nova representação mais adequada à análise ou ao modelo de aprendizado de máquina.

##5.1 Discretização

A **discretização** consiste em transformar **variáveis numéricas contínuas** (como idade, salário, temperatura etc.) em **variáveis categóricas** (como faixas etárias, níveis salariais, categorias de temperatura etc.).

Ela simplifica grandes conjuntos de valores contínuos em **intervalos gerenciáveis**, reduzindo assim a complexidade.

Um dos métodos comuns de discretização é o **binning** (agrupamento em "caixas"), técnica que divide os dados **em intervalos fixos (equal-width binning)** e baseado na frequência de elementos (**equal-frequency binning**).

````python
import pandas as pd

# Dados fictícios
dados = pd.DataFrame({
    'idade': [15, 22, 37, 45, 52, 60, 67, 80, 90]
})
````

**Vantagens:**

* **Redução de Ruído:** Ajuda a mitigar o efeito de *outliers* (valores discrepantes), colocando-os em intervalos extremos.
* **Melhora a interpretação dos dados**: facilita a compreensão dos resultados por humanos (ex: "faixa etária" é mais interpretável do que "idade exata").
* **Otimiza o processamento**: algoritmos podem computar mais rápido com menos valores distintos

**Desvantagens:**

* **Perda de Informação e Precisão**: Ao agrupar valores distintos em um mesmo intervalo (bin), perde-se a variação detalhada dentro de cada grupo. Valores que eram sutilmente diferentes, mas importantes para a predição, tornam-se idênticos
* **Perda de Poder Estatístico**: A conversão de dados contínuos para categóricos reduz a capacidade do modelo de identificar tendências lineares ou relações precisas.
* **Incompatibilidade com Algoritmos**: Alguns modelos de aprendizado de máquina, especialmente algoritmos baseados em regressão, funcionam melhor com dados contínuos e podem ter desempenho inferior com dados discretizados.
* **Descontinuidade**: A discretização transforma relações contínuas em "saltos" (jumps) entre os intervalos, o que pode representar mal a realidade de um fenômeno que muda de forma contínua

###5.1.1 Equal-width binning (intervalo fixo)

📘 **Separação Automática Conforme Quantidade de Bins**


No exemplo a seguir, **informamos a quantidade de bins**, para que a função `pd.cut()` **faça a divisão dos dados automaticamente**.

O `pd.cut()` com "equal-width binning" divide uma variável contínua em um **número predeterminado de intervalos de largura igual**, onde a função calcula o intervalo de valores (máximo - mínimo), o divide pelo número de bins especificado para obter a largura de cada bin, e então atribui cada valor a um desses intervalos.

````python
import pandas as pd
import numpy as np

valores = [5, 6, 5, 7, 12, 17, 18, 24, 33, 45, 59, 63, 70]
df = pd.DataFrame({'idade': valores})
print('Shape:', df.shape)

# 1. Equal-width binning (3 faixas de mesma largura)
df['idade_bin_width'] = pd.cut(df['idade'], bins=3,
                               labels=["Jovem","Meia-idade","Idoso"],)
print("\n=== Equal-width Binning ===")
print(df[['idade', 'idade_bin_width']])

print(f'\nFaixas de valores dos bins:')
print(df['idade_bin_width'].value_counts())

print('''
intervalo de valores de cada bin
bin 1 =  5 a 24 Jovem
bin 2 = 25 a 48 Meia-idade
bin 3 = 49 a 70 Idoso ''')

bin_series = pd.cut(df['idade'], bins=3,
                               labels=["Jovem","Meia-idade","Idoso"],)

# 1.1 Ver os intervalos de cada valor
#print(bin_series)
#print()
# 1.2 Ver apenas os intervalos únicos calculados
print(bin_series.unique())
````

A explicação para o resultado é a seguinte:

1. Calcula o valor mínimo e máximo dos dados:

```
min = 5
max = 70
```

2. Cria 3 intervalos de mesmo tamanho:

Queremos 3 bins iguais, então:

Largura do bin = ((max - min)/nº bins)
```
Range total = 70 - 5 = 65
Tamanho de cada bin = 65 / 3 = 21.66
```

3. Intervalos criados serão:

```
Bin 1: (4.935, 26.66]   # 5, 5+21.66
Bin 2: (26.66, 48.33]
Bin 3: (48.33, 70]  # ultimo valor
```

Obs: `pd.cut()` usa uma pequena margem abaixo do mínimo para garantir que o menor valor caia no primeiro intervalo, por isso o início é 4.935 (ligeiramente abaixo de 5).

<p>

**Desvantagem**

Uma desvantagem do binning de largura fixa está relacionada a situações em que os dados **não estão uniformemente distribuídos**.

Como os bins são baseados em **intervalos iguais** de valores e não na **frequência dos dados**, você pode acabar com:

* Bins com muitos dados
* Bins quase vazios ou até vazios

Com `pd.cut(valores, bins=3)`, quase todos os valores ficam no primeiro bin:

* Bin 1: 5, 6, 5, 7, 12, 17, 18, 24 → 8 valores
* Bin 2: 33, 45 → 2 valores
* Bin 3: 59, 63, 70 → 3 valores

###5.1.2 Equal-frequency binning (quantis)

O **Equal-Frequency Binning** é um tipo de discretização em que os dados são divididos em **intervalos com número (frequência) aproximadamente igual de elementos**.

Diferente do `pd.cut()` (que divide os dados por valor), o `pd.qcut()` divide por quantidade de dados — ou seja, cria faixas com o **mesmo número de observações**.

O **quantil** é um termo geral para qualquer valor que **divide um conjunto de dados ordenados em partes iguais**, enquanto o **quartil** é um tipo específico de quantil que divide os dados em exatamente quatro partes

````python
# 2. Equal-frequency binning (quantis)
df['idade_bin_quantis'] = pd.qcut(df['idade'], q=3, labels=["Baixa",
                                                            "Média",
                                                            "Alta"])
print("\n=== Equal-frequency Binning ===")
print("            (Quantis)")
print(df[['idade', 'idade_bin_quantis']])
````

De acordo com o exmeplo, o `pd.qcut()` divide os dados ordenados em 3 grupos com número igual (ou quase igual) de observações:

Como temos 13 elementos, os grupos serão aproximadamente:

* Bin 1: 4 ou 5 elementos
* Bin 2: 4 ou 5 elementos
* Bin 3: 4 elementos

Internamente, ele usa os quantis 0%, 33.3%, 66.6%, 100% para dividir os dados.

O `pd.qcut(valores, q=3)` pode gerar algo como:

```
Intervalo	     Valores atribuídos
(4.999, 7.0]	 5,   5,  6, 7
(7.0, 24.0]	     12, 17, 18, 24
(24.0, 70.0]	 33, 45, 59, 63, 70
```

**Vantagens**
* Bins equilibrados
* Evita bins vazios ou desbalanceados

**Desvantagens**
* Pode quebrar faixas lógicas (Por exemplo, pode separar idades próximas em bins diferentes)
* Limites dos bins podem parecer estranhos (Podem sair com limites como 17.0 ou 45.0, que não têm significado real)

###5.1.3 Bins especificados por limites de valores


Esta técnica de discretização divide os dados de uma coluna **em faixas predefinidas**.

Considerando que desejamos transformar a variável "idade" (que pode ir de 0 a 100) em faixas tais como:

* 0–18: Jovem
* 19–59: Adulto
* 60+: Idoso

Note que, aqui, estaremos **especificando explicitamente** os limites de cada faixa usando uma lista. Isto será feito usando a função `pd.cut()`

````python
import pandas as pd

# Dados fictícios
dados = pd.DataFrame({
    'idade': [15, 18, 19, 22, 37, 45, 52, 60, 67, 80, 90]
})

bins = [0, 18, 59, 120]
labels = ['Jovem', 'Adulto', 'Idoso']

dados['idade_categoria'] = pd.cut(dados['idade'], bins=bins, labels=labels)

print(dados)
````

###5.1.4 Discretização manual (regras do negócio)

Também chamada de **discretização supervisionada por domínio**, é um método em que desejamos transformar variáveis numéricas (como idade, renda ou temperatura) em **categorias discretas** que façam sentido no contexto do negócio ou da aplicação.

A **discretização manual** consiste em criar intervalos definidos pelo analista ou especialista de domínio, baseando-se em regras de negócio, conhecimento empírico ou critérios específicos.

Diferente das **técnicas automáticas** — como **equal-width binning** (largura igual) ou **equal-frequency binning** (frequência igual) —, essa abordagem **não depende apenas de estatísticas dos dados, mas sim de critérios interpretáveis e úteis para a tomada de decisão**.

````python
def faixa_idade(idade):
    if idade <= 12:
        return 'Criança'
    elif idade <= 17:
        return 'Adolescente'
    elif idade <= 59:
        return 'Adulto'
    else:
        return 'Idoso'

df['idade_manual'] = df['idade'].apply(faixa_idade)
print("\n=== Discretização Manual ===")
print(df[['idade', 'idade_manual']])
````
